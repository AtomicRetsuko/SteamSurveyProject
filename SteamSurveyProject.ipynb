### Looking for a Specific Data, Find Duplicates

```python
#To see games by editors, just type 'Publishers' as a parameter, like Bethesda Game Studios
#To seek a certain game, use name as a Parameter and then the name of your game
#Search Algorythm, looking for a certain data like "SELECT * FROM Games WHERE name = 'Fallout 4'"

# Load the CSV into PySpark DataFrame
df = spark.read.csv('Games.csv', header=True, inferSchema=True)

#Asking the User about the Precise Data he wants to search within a certain Paramter
f_Param = input("Please enter The Parameter you want to sear in :")
f_data = input(f"Please Enter the data you want to search with the {f_Param} parameter")

#Initializing a GLOBAL Variable to treat the loaded Data like a SQL DataBase
df.createOrReplaceTempView("GLOBAL")

#Try to Display the precise Data We're Looking for, as you can see, it also works to find Duplicates
try:
    query = f"SELECT * FROM GLOBAL WHERE `{f_Param}` = '{f_data}'"
    spark.sql(query).show()
except Exception as e: #saves the exception parameter in e
    print(f"Error executing query: {e}")
```

### Looking for How Much Games are Produced by a Specific Publisher

```python
df_spark = spark.read.csv('Games.csv', header=True, inferSchema=True)

my_column = 'Publishers'
ChosenPublisher = input("Please Enter a Publisher : \n")


nb_ocurrence = df_spark.filter(df_spark[my_column] == ChosenPublisher).count() 

print(f"There's {nb_ocurrence} Games made by {ChosenPublisher}")
```

### Ranking Publishers by games produced 

```python
df_spark = spark.read.csv('Games.csv', header=True, inferSchema=True)

my_column = 'Publishers'

counteuh = df_spark.groupBy(my_column).count().orderBy('count', ascending=False)

counteuh.show(n=1000, truncate=False)

```

### Games out by Year

```python
#Games out by Year

df_spark = spark.read.csv('Games.csv', header=True, inferSchema=True)

my_column = 'Release date'

counteuh = df_spark.groupBy(my_column).count().orderBy('count', ascending=False)

counteuh.show(n=1000, truncate=False)
```
